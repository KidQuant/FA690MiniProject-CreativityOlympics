```python
import time
import numpy as np
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import warnings
warnings.filterwarnings('ignore')
```

```python
user_input_job_title = input('Enter the job title you want to search for: ').split()
print(user_input_job_title)
```

```python
b = []
for i in user_input_job_title:
    x = i.split()
    y = '%20'.join(x)
    b.append(y)

job_title = '+'.join(b)
print(job_title)
```

```python
link = f"https://in.linkedin.com/jobs/search?keywords={job_title}&location=India&locationId=&geoId=102713980&f_TPR=r604800&position=1&pageNum=0"
print(link)
```

```python
driver = webdriver.Chrome()
driver.maximize_window()

driver.get(link)
driver.implicitly_wait(10)
```

```python
for i in range(0,2):
    driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
    time.sleep(5)

    try:
        x = driver.find_element(by=By.CSS_SELECTOR, value="button[aria-label='See more jobs']").click()
        time.sleep(3)
    except:
        pass
```

```python
company_name = []

try:
    company = driver.find_elements(by=By.CSS_SELECTOR, value='h4[class="base-search-card__subtitle"]')
    for i in company:
        company_name.append(i.text)

except:
    pass

print(company_name)
```
